### Cleaning

```python
# dropping irrelavant columns
df_foodprices = df_foodprices.drop(columns=['market_id', 'priceflag', 'currency'])
df_poverty = df_poverty.drop(columns=['location_code', 'has_hrp', 'in_gho', 'admin1_code', 'admin1_name', 'admin_level'])

print("shape of food prices data: ", df_foodprices.shape)
print("shape of poverty data: ", df_poverty.shape)
```

```python
# missigng values
print(df_foodprices.isna().sum())
print()
print(df_poverty.isna().sum())
```

```python
df_foodprices = df_foodprices.dropna(subset=['admin1', 'admin2', 'latitude', 'longitude'])
df_poverty = df_poverty.dropna(subset='provider_admin1_name')

print(df_foodprices.shape)
print(df_poverty.shape)
```

```python
# changing datatypes
df_foodprices['date'] = pd.to_datetime(df_foodprices['date'])

df_poverty['reference_period_start'] = pd.to_datetime(df_poverty['reference_period_start'])
df_poverty['reference_period_end'] = pd.to_datetime(df_poverty['reference_period_end'])
```

```python
# handling duplicates
df_foodprices = df_foodprices.duplicated().sum()
```

```python
df_poverty.duplicated().sum()
```

```python
# whitespace check
print(df_foodprices['admin1'].unique()[:5])
print(df_poverty['provider_admin1_name'].unique()[:5])
```

```python
# filtering irrelavant rows
print(df_foodprices['pricetype'].unique())
```

```python
# final check
print("Food prices shape:", df_foodprices.shape)
print("Poverty shape:", df_poverty.shape)
print()
print(df_foodprices.describe())
print()
print(df_poverty.describe())
```

```python
# saving cleaned df
df_foodprices.to_csv('../data/food_prices_cleaned.csv', index= False)
df_poverty.to_csv('../data/poverty_cleaned.csv', index=False)

print("saved cleaned df")
```
