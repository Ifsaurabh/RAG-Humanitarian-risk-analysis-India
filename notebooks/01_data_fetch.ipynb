## Loading data and Creating dataframe

```python
url = "https://data.humdata.org/dataset/wfp-food-prices-for-india/resource/3b1ff071-6b01-4998-aa62-2f3efb5d4888/download/wfp_food_prices_ind.csv"
```

```python
# libraries imports
import os
import requests
import pandas as pd
import io
```

```python
# loading data
headers = {"User-Agent": "Mozilla/5.0"}
response = requests.get(url, headers=headers)
print(response.status_code)
```

```python
df_foodprices = pd.read_csv(io.StringIO(response.text))
df_foodprices.head()
```
